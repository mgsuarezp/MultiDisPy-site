# MultiDisPy

<p class="lead">How does biodiversity change across space and through time?</p>

Two assemblages can contain the same number of species yet differ in the range of forms, ecological roles, and taxonomic lineages they represent.

![MultiDisPy connects different dimensions of biodiversity: taxonomic, morphological, ecological and functional.](images/dimensions.png){width=40%}

MultiDisPy is a Python package in development for exploring these dimensions together. It brings morphological, taxonomic, functional, and ecological approaches into a common framework for comparing living and fossil assemblages. A phylogenetic component is planned.
